Question 1 : In simple words, what are React Hooks? Create a small component that
shows a button. When the button is clicked, toggle the text between “On” and “Off”.


Answer 01            
    React Hooks are built-in functions introduced in React 16.8 that allow developers to use state and other React lifecycle features inside functional components without writing class components.

Before hooks, functional components were stateless ("dumb" components). With hooks like useState and useEffect, functional components can manage local data, handle side effects, and share reusable logic cleanly and efficiently.

```python
import React, { useState } from 'react';

function ToggleButton() {
  // Declare a state variable 'isOn' initialized to false
  const [isOn, setIsOn] = useState(false);

  // Function to flip the boolean state
  const handleToggle = () => {
    setIsOn((prevState) => !prevState);
  };

  return (
    <div>
      <button onClick={handleToggle}>
        {isOn ? 'On' : 'Off'}
      </button>
    </div>
  );
}

export default ToggleButton;
```

Question 2 : Make a counter with a button. Whenever the count changes, log “Count
changed” using useEffect.


```python
import React, { useState, useEffect } from 'react';

function Counter() {
  // 1. Initialize count state to 0
  const [count, setCount] = useState(0);

  // 2. useEffect runs whenever the 'count' dependency changes
  useEffect(() => {
    console.log("Count changed");
  }, [count]);

  return (
    <div>
      <h2>Count: {count}</h2>
      <button onClick={() => setCount(prevCount => prevCount + 1)}>
        Increment
      </button>
    </div>
  );
}

export default Counter;
```

Question  3  
: Write a component that fetches users from
https://jsonplaceholder.typicode.com/users only once on mount and
shows how many users were loaded.


Answer 03  


Here is a component that fetches data from the provided API when the component first renders (mounts) and displays the total number of users loaded.

```python
import React, { useState, useEffect } from 'react';

function UserFetcher() {
  // 1. State to store the fetched users
  const [users, setUsers] = useState([]);

  // 2. useEffect with an empty dependency array runs ONLY ONCE when mounted
  useEffect(() => {
    const fetchUsers = async () => {
      try {
        const response = await fetch('https://jsonplaceholder.typicode.com/users');
        const data = await response.json();
        // Update the state with the fetched data
        setUsers(data);
      } catch (error) {
        console.error("Error fetching data:", error);
      }
    };

    fetchUsers();
  }, []); // The empty array [] is the key to running this just once

  return (
    <div>
      <h2>Number of users loaded: {users.length}</h2>
    </div>
  );
}

export default UserFetcher;
```

Question 4 : Create a component that starts a timer with setInterval to increase a
“seconds” counter every second. Stop the timer when the component unmounts.


```python
import React, { useState, useEffect } from 'react';

function Timer() {
  // 1. State variable to track seconds passed
  const [seconds, setSeconds] = useState(0);

  useEffect(() => {
    // 2. Set up the interval timer to tick every 1000ms (1 second)
    const intervalId = setInterval(() => {
      setSeconds((prevSeconds) => prevSeconds + 1);
    }, 1000);

    // 3. Cleanup function: runs when the component unmounts
    return () => {
      clearInterval(intervalId);
    };
  }, []); // Empty dependency array ensures the interval is created only once on mount

  return (
    <div>
      <h2>Timer: {seconds} seconds</h2>
    </div>
  );
}

export default Timer;
```

Question 5 : Build a small form with a text input. If the input is empty, show “Please
type your name”. If it has text, show “Hello, <name>”.


```python
import React, { useState } from 'react';

function NameForm() {
  // 1. State to track the input text value
  const [name, setName] = useState('');

  return (
    <div>
      <input
        type="text"
        placeholder="Enter your name"
        value={name}
        onChange={(e) => setName(e.target.value)}
      />

      {/* 2. Conditional rendering based on whether input is empty */}
      <p>
        {name.trim() === '' ? 'Please type your name' : `Hello, ${name}`}
      </p>
    </div>
  );
}

export default NameForm;
```

Question 6 : You have a theme value (“light” or “dark”) that is needed by a deep child
button. Show how to pass it without sending theme through every parent prop.


```python
import React, { createContext, useContext, useState } from 'react';

// 1. Create a Context for the theme
const ThemeContext = createContext();

// 2. Deep Child Component consuming the context
function ThemedButton() {
  // Access the theme value directly without props
  const theme = useContext(ThemeContext);

  const style = {
    backgroundColor: theme === 'dark' ? '#333' : '#FFF',
    color: theme === 'dark' ? '#FFF' : '#000',
    padding: '8px 16px',
    border: '1px solid #ccc',
    borderRadius: '4px',
    cursor: 'pointer'
  };

  return <button style={style}>Current Theme: {theme}</button>;
}

// 3. Intermediate Parent Component (passes NO props down)
function Toolbar() {
  return (
    <div>
      <p>Intermediate component (no props passed here)</p>
      <ThemedButton />
    </div>
  );
}

// 4. Top-level App Component providing the context
export default function App() {
  const [theme, setTheme] = useState('dark');

  const toggleTheme = () => {
    setTheme((prevTheme) => (prevTheme === 'light' ? 'dark' : 'light'));
  };

  return (
    // Wrap descendants in the Provider and pass the current value
    <ThemeContext.Provider value={theme}>
      <div>
        <button onClick={toggleTheme}>Toggle Theme</button>
        <hr />
        <Toolbar />
      </div>
    </ThemeContext.Provider>
  );
}
```

Question 7 : Make a counter with two buttons: “+1” and “-1”. Use useReducer instead
of useState.


```python
import React, { useReducer } from 'react';

// 1. Define the initial state
const initialState = { count: 0 };

// 2. Define the reducer function
function counterReducer(state, action) {
  switch (action.type) {
    case 'INCREMENT':
      return { count: state.count + 1 };
    case 'DECREMENT':
      return { count: state.count - 1 };
    default:
      return state;
  }
}

function Counter() {
  // 3. Initialize useReducer with the reducer function and initial state
  const [state, dispatch] = useReducer(counterReducer, initialState);

  return (
    <div>
      <h2>Count: {state.count}</h2>
      <button onClick={() => dispatch({ type: 'INCREMENT' })}>+1</button>
      <button onClick={() => dispatch({ type: 'DECREMENT' })}>-1</button>
    </div>
  );
}

export default Counter;
```

Question 8 : Given an array nums = [1,2,3,4] and a filter value min, show how to
compute a filtered list only when nums or min change. Also pass a stable click handler
to a child list item.

Answer 08

To compute the filtered list only when nums or min change, use useMemo. To ensure the child component receives a stable function reference across re-renders, wrap the handler in useCallback and wrap the child item in React.memo.

```python
import React, { useState, useMemo, useCallback } from 'react';

// 1. Child component wrapped in React.memo to skip re-renders if props don't change
const ListItem = React.memo(({ value, onClick }) => {
  console.log(`Rendering ListItem: ${value}`);
  return (
    <li onClick={() => onClick(value)} style={{ cursor: 'pointer', margin: '4px 0' }}>
      Item: {value} (click me)
    </li>
  );
});

export default function NumberFilterList() {
  const [nums, setNums] = useState([1, 2, 3, 4]);
  const [min, setMin] = useState(2);
  const [unrelatedState, setUnrelatedState] = useState(0);

  // 2. Compute filtered list only when 'nums' or 'min' change
  const filteredNums = useMemo(() => {
    console.log('Recalculating filtered list...');
    return nums.filter((num) => num >= min);
  }, [nums, min]);

  // 3. Stable click handler that maintains the same reference across re-renders
  const handleClick = useCallback((item) => {
    console.log(`Clicked item: ${item}`);
  }, []); // Empty dependencies because it doesn't rely on outer state variables

  return (
    <div>
      <h3>Filtered Numbers (min: {min})</h3>
      <ul>
        {filteredNums.map((num) => (
          <ListItem key={num} value={num} onClick={handleClick} />
        ))}
      </ul>

      {/* Controls */}
      <button onClick={() => setMin((prev) => prev + 1)}>Increase Min</button>
      <button onClick={() => setNums((prev) => [...prev, prev.length + 1])}>Add Number</button>
      <button onClick={() => setUnrelatedState((prev) => prev + 1)}>
        Trigger Unrelated Re-render ({unrelatedState})
      </button>
    </div>
  );
}
```

Question 9 : Create a controlled input for email. As the user types, show “Email:
<value>”. Why do we keep the input’s value in state?

```python
import React, { useState } from 'react';

function ControlledEmailInput() {
  // 1. Store the input value in React state
  const [email, setEmail] = useState('');

  const handleChange = (e) => {
    setEmail(e.target.value);
  };

  return (
    <div>
      <label htmlFor="email-field">Enter Email: </label>
      <input
        id="email-field"
        type="email"
        placeholder="example@mail.com"
        value={email}
        onChange={handleChange}
      />

      {/* Real-time output displaying what the user types */}
      <p>Email: {email}</p>
    </div>
  );
}

export default ControlledEmailInput;
```

Question 10 : Make a button that focuses an input using a ref. Return adjacent
elements without extra wrapper nodes. In one sentence, say what React’s update
process does.


Answer 10

To focus the input element, use the useRef hook. To return adjacent JSX elements without adding an extra wrapper node (like an unnecessary <div>), wrap them in a React Fragment (<> ... </>)

```python
import React, { useRef } from 'react';

function AutoFocusInput() {
  // 1. Create a ref to hold the DOM node reference
  const inputRef = useRef(null);

  // 2. Click handler to imperatively focus the input element
  const handleFocus = () => {
    inputRef.current.focus();
  };

  return (
    // Fragment returns adjacent elements without adding an extra wrapper node in the DOM
    <>
      <input
        ref={inputRef}
        type="text"
        placeholder="Click button to focus me"
      />
      <button onClick={handleFocus}>
        Focus Input
      </button>
    </>
  );
}

export default AutoFocusInput;
```
